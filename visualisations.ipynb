```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Change only this line when you use a different dataset
FILE = "indian_sales_practice.csv"

df = pd.read_csv(FILE)
df["order_date"] = pd.to_datetime(df["order_date"])
df.head()
```

```python
# Chart 1: LINE chart - total sales for each month
monthly = df.groupby(df["order_date"].dt.to_period("M"))["sales"].sum()

plt.figure(figsize=(9, 4))
plt.plot(monthly.index.astype(str), monthly.values, marker="o")
plt.title("Monthly Sales in 2024")
plt.xlabel("Month")
plt.ylabel("Total Sales (INR)")
plt.xticks(rotation=45)
plt.savefig("chart1_line.png", bbox_inches="tight")
plt.show()
```

```python
# Chart 2: BAR chart - total sales for each region
region_sales = df.groupby("region")["sales"].sum().sort_values(ascending=False)

plt.figure(figsize=(7, 4))
plt.bar(region_sales.index, region_sales.values, color="steelblue")
for i, v in enumerate(region_sales.values):
    plt.text(i, v, f"{v:,.0f}", ha="center", va="bottom")
plt.title("Total Sales by Region")
plt.xlabel("Region")
plt.ylabel("Total Sales (INR)")
plt.savefig("chart2_bar.png", bbox_inches="tight")
plt.show()
```

```python
# Chart 3: HISTOGRAM - how are the sales values spread?
sns.set_theme(style="whitegrid")

plt.figure(figsize=(7, 4))
sns.histplot(df["sales"], bins=20)
plt.title("Distribution of Order Sales")
plt.xlabel("Sales per Order (INR)")
plt.ylabel("Number of Orders")
plt.savefig("chart3_histogram.png", bbox_inches="tight")
plt.show()
```

```python
# Chart 4: SCATTER plot - sales vs profit
plt.figure(figsize=(7, 4))
sns.scatterplot(data=df, x="sales", y="profit", hue="category")
plt.axhline(0, color="red", linestyle="--")
plt.text(df["sales"].max() * 0.55, 3000, "Below the red line = loss", color="red")
plt.title("Sales vs Profit")
plt.xlabel("Sales per Order (INR)")
plt.ylabel("Profit (INR)")
plt.savefig("chart4_scatter.png", bbox_inches="tight")
plt.show()
```

```python
# Chart 5: BOX plot - sales for each category
plt.figure(figsize=(7, 4))
sns.boxplot(data=df, x="category", y="sales")
plt.title("Order Sales by Category")
plt.xlabel("Category")
plt.ylabel("Sales per Order (INR)")
plt.savefig("chart5_box.png", bbox_inches="tight")
plt.show()
```
