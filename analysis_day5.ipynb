```python
df_test = df.copy()
df_test.loc[0:4, "profit"] = None
print("Missing before:", df_test["profit"].isnull().sum())

df_test["profit"] = df_test["profit"].fillna(df_test["profit"].mean())
print("Missing after:", df_test["profit"].isnull().sum())
```

```python
df.groupby("region")["sales"].sum()
```

```python
summary = df.groupby("region").agg(
    orders=("order_id", "count"),
    total_sales=("sales", "sum"),
    avg_profit=("profit", "mean")
).round(2).sort_values("total_sales", ascending=False)

summary
```
