```python
import os
import pandas as pd

# Find the Superstore file in this folder
files = [f for f in os.listdir() if f.lower().endswith(".csv") and "superstore" in f.lower()]
print("Found:", files)

FILE = files[0]

# Some Superstore files need a different encoding
try:
    df = pd.read_csv(FILE)
except UnicodeDecodeError:
    df = pd.read_csv(FILE, encoding="latin-1")

print("Loaded:", FILE)
print("Shape:", df.shape)
df.head()
```

```python
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
df.info()
```

```python
missing = df.isnull().sum()
missing[missing > 0]
```

```python
print("Duplicate rows:", df.duplicated().sum())
```

```python
df["Order Date"] = pd.to_datetime(df["Order Date"], format="mixed")
df["Ship Date"] = pd.to_datetime(df["Ship Date"], format="mixed")

print("First order:", df["Order Date"].min())
print("Last order:", df["Order Date"].max())
print(df.columns.tolist())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
num_cols = ["Sales", "Quantity", "Discount", "Profit"]

fig, axes = plt.subplots(2, 2, figsize=(11, 8))
for ax, col in zip(axes.flatten(), num_cols):
    sns.histplot(df[col], bins=40, ax=ax)
    ax.set_title(f"Distribution of {col}")
plt.tight_layout()
plt.savefig("eda_distributions.png", bbox_inches="tight")
plt.show()
```

```python
# Keep only the real measurement columns (not the ID-like numbers)
num_df = df.select_dtypes(include="number").drop(columns=["Row ID", "Postal Code"])

corr = num_df.corr()

plt.figure(figsize=(6, 5))
sns.heatmap(corr, annot=True, fmt=".2f", cmap="coolwarm", vmin=-1, vmax=1)
plt.title("Correlation Heatmap")
plt.savefig("eda_heatmap.png", bbox_inches="tight")
plt.show()
```

```python
cat = df.groupby("Category").agg(
    total_sales=("Sales", "sum"),
    total_profit=("Profit", "sum")
).round(0)
cat["profit_margin_%"] = (cat["total_profit"] / cat["total_sales"] * 100).round(1)
print(cat)

plt.figure(figsize=(6, 4))
ax = sns.barplot(x=cat.index, y=cat["total_profit"])
for c in ax.containers:
    ax.bar_label(c, fmt="%.0f")
plt.title("Total Profit by Category")
plt.xlabel("Category")
plt.ylabel("Total Profit")
plt.savefig("insight1_category_profit.png", bbox_inches="tight")
plt.show()
```

```python
df["Discount Band"] = pd.cut(
    df["Discount"],
    bins=[-0.01, 0, 0.2, 0.4, 1.0],
    labels=["No discount", "1-20%", "21-40%", "Above 40%"]
)

disc = df.groupby("Discount Band", observed=True)["Profit"].agg(
    orders="count", avg_profit="mean", total_profit="sum"
).round(2)
print(disc)

plt.figure(figsize=(6, 4))
sns.barplot(x=disc.index.astype(str), y=disc["avg_profit"])
plt.axhline(0, color="red", linestyle="--")
plt.title("Average Profit per Order by Discount Level")
plt.xlabel("Discount Band")
plt.ylabel("Average Profit")
plt.savefig("insight2_discount_profit.png", bbox_inches="tight")
plt.show()
```
