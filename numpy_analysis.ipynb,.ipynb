```python
import numpy as np

# 1D array: one row of numbers
a1 = np.array([10, 20, 30, 40, 50])

# 2D array: a table with 2 rows and 3 columns
a2 = np.array([[1, 2, 3],
               [4, 5, 6]])

# 3D array: 2 tables, each with 3 rows and 4 columns
a3 = np.arange(24).reshape(2, 3, 4)

for name, arr in [("1D", a1), ("2D", a2), ("3D", a3)]:
    print(name, "| shape:", arr.shape, "| dimensions:", arr.ndim, "| total items:", arr.size)

print("a2 row 1, column 2:", a2[1, 2])
print("a2 transposed shape:", a2.T.shape)
```

```python
import os
import pandas as pd

files = [f for f in os.listdir() if f.lower().endswith(".csv") and "superstore" in f.lower()]
FILE = files[0]

try:
    df = pd.read_csv(FILE)
except UnicodeDecodeError:
    df = pd.read_csv(FILE, encoding="latin-1")

sales = df["Sales"].to_numpy()
profit = df["Profit"].to_numpy()
discount = df["Discount"].to_numpy()

print("Rows:", len(sales))
print("First 5 sales:", sales[:5])
```

```python
# Profit margin (%) for every order, with no loop
margin_pct = profit / sales * 100
print("First 5 margins (%):", margin_pct[:5].round(1))

# Put 3 columns side by side into one 2D array
data = np.column_stack([sales, profit, discount])
print("Shape:", data.shape)

# axis=0 means "go down the rows", so you get one answer for each column
print("Mean of Sales, Profit, Discount:", data.mean(axis=0).round(2))
print("Sum of Sales, Profit, Discount:", data.sum(axis=0).round(0))
```

```python
# A mask is a True/False list. True means "keep this row".
loss_mask = profit < 0
print("Loss-making orders:", loss_mask.sum())
print("Percent of all orders:", round(loss_mask.mean() * 100, 1), "%")
print("Total sales of loss orders:", round(sales[loss_mask].sum(), 0))
print("Total loss:", round(profit[loss_mask].sum(), 0))

# Two conditions: high discount AND loss
combo_mask = (discount > 0.3) & (profit < 0)
print("High-discount loss orders:", combo_mask.sum())
```

```python
# Positions of the 5 biggest sales
top5 = np.argsort(sales)[-5:][::-1]
print("Top 5 sales:", sales[top5])
df.iloc[top5][["Order ID", "Sales"]]
```
